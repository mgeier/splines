# An Example Curve

Example from Dam et al 1998, used again in Haarbach et al 2018.

```python
from splines.quaternion import UnitQuaternion, CatmullRom
```

Note that un-normalized rotation axes are given (which is fine because `from_axis_angle()` normalizes).

```python
rotations = [
    UnitQuaternion.from_axis_angle((1, 3, 0), 1),
    UnitQuaternion.from_axis_angle((-1, 0, 0), 1.9),
    UnitQuaternion.from_axis_angle((-2, 1, 0), 0),
    UnitQuaternion.from_axis_angle((3, 4, 0), -2),
    UnitQuaternion.from_axis_angle((-1, 4, 0), -1),
    UnitQuaternion.from_axis_angle((1, 3, 0), 1),
    # Haarbach et al. (2018) uses this instead of the last one:
    #UnitQuaternion.from_axis_angle((2, 3, 0), 1),
]
```

The rotations in the Dam paper cycle back to the same orientation value,
but the images in the paper don't show closed curves.
So something seems wrong there.

```python
s = CatmullRom(rotations)
```
